# 📘 Day 16 — Cards, KPIs, and Conditional Formatting

# ⭐ 1. Overview
This file focuses on **highlighting key metrics using card visuals, KPIs, and applying conditional formatting** to tables and charts.  
These features improve **dashboard readability and data storytelling**.

# ⭐ 2. Card Visuals
- **Purpose:** Display a single key metric (e.g., Total Revenue, Total Orders)  
- **Steps to create:**  
  1. Insert → Card  
  2. Drag desired measure or column to **Values**  
  3. Format font, background, and size in the **Visualizations pane**  
- **Example:** Total Revenue, Total Quantity, Unique Products

# ⭐ 3. KPI Visuals
- **Purpose:** Show performance against a target with trend indicator  
- **Components:**  
  - **Indicator:** Actual value  
  - **Trend axis:** Time dimension (e.g., Month)  
  - **Target value:** Set benchmark for comparison  
- **Steps to create:**  
  1. Insert → KPI  
  2. Drag measure to **Indicator**  
  3. Drag Date to **Trend axis**  
  4. Drag target measure to **Target goal**  
- **Example:** Total Revenue vs Target Revenue over months

# ⭐ 4. Conditional Formatting
- **Purpose:** Highlight important values or patterns in tables/matrix  
- **Options:** Background color, font color, data bars, icons  
- **Steps to apply:**  
  1. Select table or matrix visual  
  2. In **Values field**, click the down arrow → Conditional formatting  
  3. Choose formatting type (e.g., Background color, Font color)  
  4. Set rules (e.g., Total Revenue > 500 → Green; ≤ 500 → Red)  
- **Example:** Highlight high revenue rows in green, low in red

# ⭐ 5. Example Scenario
1. **Card Visuals:** Show Total Revenue, Total Quantity, Unique Products  
2. **KPI Visual:** Track Total Revenue against Target Revenue monthly  
3. **Table Visual with Conditional Formatting:**  
   - Columns: Product, Category, Total Revenue  
   - Apply **Background color** based on revenue thresholds  
4. **Dashboard Layout:** Place cards at top, KPI below, table with conditional formatting on the side

# ⭐ 6. Key Takeaways
- Card visuals summarize **key metrics** at a glance  
- KPI visuals help track **performance against targets**  
- Conditional formatting improves **visual insight and quick analysis**  
- Combining these features enhances dashboard storytelling and decision-making

# ⭐ 7. Interview Questions

**Q1: What is a card visual used for in Power BI?**  
To display a single key metric clearly on the dashboard.

**Q2: How do you create a KPI visual?**  
Insert → KPI → Set Indicator (measure), Trend axis (time), Target goal (measure).

**Q3: What are the components of a KPI visual?**  
Indicator, Trend axis, Target goal.

**Q4: How does conditional formatting improve a dashboard?**  
By highlighting important values, trends, and patterns for better readability.

**Q5: What types of conditional formatting are available?**  
Background color, font color, data bars, and icons.

**Q6: How do you apply conditional formatting to a table or matrix?**  
Select table/matrix → Values field → Conditional formatting → Choose type → Set rules.

**Q7: Can you use conditional formatting in charts?**  
Yes, some charts like bar or column allow data-driven color formatting.

**Q8: Why are cards and KPIs important in dashboards?**  
They emphasize critical metrics and track performance at a glance.

**Q9: How do you set a target value in KPI visual?**  
Drag a measure representing the target to the Target goal field.

**Q10: What is the difference between card and KPI visuals?**  
Card shows a single value; KPI shows value vs target along with a trend indicator.
