# 📘 Day 14 — Practice & Mini Exercise (DAX & Time Intelligence)

# ⭐ 1. Overview
Day 14 focuses on **practicing DAX calculations and time intelligence functions**.  
The goal is to create a simple sales dashboard using measures, aggregation, logical, filter, and time intelligence functions.

# ⭐ 2. Step-by-Step Workflow

1. **Import Dataset**  
   - Load Sales table (SalesID, ProductID, Quantity, Price, OrderDate)  
   - Load Product table (ProductID, Product, Category)  
   - Load Date table (Date, Month, Year, Quarter)

2. **Verify Relationships**  
   - Go to Model view → Check relationships:  
     - Sales[ProductID] → Product[ProductID] (One-to-Many)  
     - Sales[OrderDate] → Date[Date] (One-to-Many)  

3. **Create Basic Measures**  
   - Total Revenue: `Total Revenue := SUMX(Sales, Sales[Quantity] * Sales[Price])`  
   - Total Quantity: `Total Quantity := SUM(Sales[Quantity])`  
   - Unique Products: `Unique Products := DISTINCTCOUNT(Sales[ProductID])`  

4. **Create Conditional Measure**  
   - High Sales Flag: `High Sales := IF([Total Revenue] > 500, "Yes", "No")`  

5. **Apply Filter Functions**  
   - Stationery Sales: `Stationery Sales := CALCULATE([Total Revenue], Product[Category] = "Stationery")`  
   - Large Orders: `Large Orders := CALCULATE([Total Revenue], FILTER(Sales, Sales[TotalRevenue] > 500))`  

6. **Apply Time Intelligence Functions**  
   - YTD Sales: `YTD Sales := TOTALYTD([Total Revenue], Sales[OrderDate])`  
   - MTD Sales: `MTD Sales := TOTALMTD([Total Revenue], Sales[OrderDate])`  
   - QTD Sales: `QTD Sales := TOTALQTD([Total Revenue], Sales[OrderDate])`  
   - Sales Last Year: `Sales Last Year := CALCULATE([Total Revenue], SAMEPERIODLASTYEAR(Sales[OrderDate]))`  

7. **Create a Simple Dashboard**  
   - Insert a **Card visual** for Total Revenue, Total Quantity, Unique Products  
   - Insert a **Clustered Column chart** for Revenue by Product Category  
   - Insert a **Line chart** for YTD Sales trend  
   - Insert a **Slicer** for Year or Month to filter visuals  

8. **Document Workflow**  
   - Step 1: Import datasets  
   - Step 2: Verify relationships  
   - Step 3: Create basic measures  
   - Step 4: Create conditional measures  
   - Step 5: Apply filter functions  
   - Step 6: Apply time intelligence functions  
   - Step 7: Build visuals for dashboard  
   - Step 8: Test slicers and verify calculations  

# ⭐ 3. Key Takeaways
- Practicing DAX and time intelligence reinforces understanding of **row vs filter context**  
- Measures are preferred for aggregations and dynamic calculations  
- Time intelligence functions allow trend analysis and comparison with previous periods  
- Building a dashboard integrates multiple concepts for end-to-end reporting

# ⭐ 4. Interview Questions
**Q1: How do you calculate total revenue in Power BI using DAX?**  
Use SUMX to multiply Quantity by Price for each row: `Total Revenue := SUMX(Sales, Sales[Quantity] * Sales[Price])`

**Q2: How do you count unique products sold?**  
Use DISTINCTCOUNT on ProductID: `Unique Products := DISTINCTCOUNT(Sales[ProductID])`

**Q3: How do you create a conditional measure for high sales?**  
Use IF function: `High Sales := IF([Total Revenue] > 500, "Yes", "No")`

**Q4: How do you calculate category-specific sales?**  
Use CALCULATE with a filter: `Stationery Sales := CALCULATE([Total Revenue], Product[Category] = "Stationery")`

**Q5: How do you calculate YTD sales?**  
Use TOTALYTD function: `YTD Sales := TOTALYTD([Total Revenue], Sales[OrderDate])`

**Q6: How do you calculate sales for the same period last year?**  
Use SAMEPERIODLASTYEAR in CALCULATE: `Sales Last Year := CALCULATE([Total Revenue], SAMEPERIODLASTYEAR(Sales[OrderDate]))`

**Q7: Why use slicers in a dashboard?**  
To dynamically filter visuals based on selected criteria like Year, Month, or Product.

**Q8: Why are measures preferred over calculated columns in dashboards?**  
Measures are dynamic, memory-efficient, and respect filter context, making them ideal for aggregated visuals.

**Q9: How do you apply filters within a measure?**  
Use CALCULATE with column filters or FILTER function for complex conditions.

**Q10: What is the importance of verifying relationships before creating measures?**  
Relationships ensure that related table columns are correctly connected, which is essential for accurate calculations and DAX functions.
