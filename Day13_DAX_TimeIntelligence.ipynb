# 📘 Day 13 — Time Intelligence in DAX

# ⭐ 1. Overview
Time intelligence functions in DAX allow you to **perform calculations over time**, such as year-to-date, month-to-date, quarter-to-date, or comparing with previous periods.  
They are essential for analyzing trends and growth in Power BI.

# ⭐ 2. Requirements for Time Intelligence
- A **Date column** in your dataset  
- A **Date table** (calendar table) marked as a **date table** in Power BI  
- Proper relationship between Date table and fact table

# ⭐ 3. Common Time Intelligence Functions

- **TOTALYTD:** Year-to-date total  
  Syntax: TOTALYTD(Expression, DatesColumn)  
  Example: `YTD Sales := TOTALYTD([Total Revenue], Sales[OrderDate])`

- **TOTALMTD:** Month-to-date total  
  Syntax: TOTALMTD(Expression, DatesColumn)  
  Example: `MTD Sales := TOTALMTD([Total Revenue], Sales[OrderDate])`

- **TOTALQTD:** Quarter-to-date total  
  Syntax: TOTALQTD(Expression, DatesColumn)  
  Example: `QTD Sales := TOTALQTD([Total Revenue], Sales[OrderDate])`

- **SAMEPERIODLASTYEAR:** Compare same period last year  
  Syntax: SAMEPERIODLASTYEAR(DatesColumn)  
  Example: `Sales Last Year := CALCULATE([Total Revenue], SAMEPERIODLASTYEAR(Sales[OrderDate]))`

- **DATESYTD / DATESMTD / DATESQTD:** Returns a table of dates for YTD, MTD, QTD  
  Example: `Dates YTD := DATESYTD(DateTable[Date])`

# ⭐ 4. Example Scenario
Sales table: SalesID, ProductID, Quantity, Price, OrderDate  

1. Total Revenue measure: `Total Revenue := SUMX(Sales, Sales[Quantity] * Sales[Price])`  
2. Year-to-date sales: `YTD Sales := TOTALYTD([Total Revenue], Sales[OrderDate])`  
3. Month-to-date sales: `MTD Sales := TOTALMTD([Total Revenue], Sales[OrderDate])`  
4. Quarter-to-date sales: `QTD Sales := TOTALQTD([Total Revenue], Sales[OrderDate])`  
5. Sales for same period last year: `Sales Last Year := CALCULATE([Total Revenue], SAMEPERIODLASTYEAR(Sales[OrderDate]))`

# ⭐ 5. Key Takeaways
- Time intelligence functions allow **dynamic calculations over time**  
- Always use a **Date table** for accurate results  
- YTD, MTD, QTD help analyze growth and trends  
- SAMEPERIODLASTYEAR is useful for comparing periods  
- Proper relationships and marked date tables are essential

# ⭐ 6. Interview Questions
**Q1: What are time intelligence functions in DAX?**  
Functions that allow calculations over time such as YTD, MTD, QTD, and period comparisons.

**Q2: What is required to use time intelligence functions correctly?**  
A Date column, a Date table marked as a date table, and a relationship between Date table and fact table.

**Q3: What does TOTALYTD do?**  
Calculates the year-to-date total for a measure.

**Q4: What does SAMEPERIODLASTYEAR do?**  
Returns values for the same period in the previous year for comparison.

**Q5: How do TOTALMTD and TOTALQTD differ from TOTALYTD?**  
They calculate totals for month-to-date and quarter-to-date respectively instead of year-to-date.

**Q6: What is the purpose of DATESYTD, DATESMTD, DATESQTD?**  
They return a table of dates corresponding to YTD, MTD, or QTD for use in calculations.

**Q7: Can time intelligence functions be used without a Date table?**  
Technically yes, but results may be inaccurate; a Date table is strongly recommended.

**Q8: Why do we mark a table as a date table in Power BI?**  
So DAX knows which column represents continuous dates for time intelligence calculations.

**Q9: How can you calculate last year's sales for a specific period?**  
Use SAMEPERIODLASTYEAR within CALCULATE, e.g., `CALCULATE([Total Revenue], SAMEPERIODLASTYEAR(Sales[OrderDate]))`.

**Q10: Why is time intelligence important in business reporting?**  
It helps analyze trends, growth, and performance over time for better decision-making.
