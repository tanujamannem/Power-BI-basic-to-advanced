# 📘 Day 17 — Advanced Visuals

# ⭐ 1. Overview
This file focuses on **advanced visuals in Power BI** such as combo charts, waterfall, funnel, gauge, and maps.  
These visuals help present **complex data and insights** clearly.

# ⭐ 2. Combo Charts
- **Purpose:** Combine column and line chart to show two measures together  
- **Steps to create:**  
  1. Insert → Combo Chart → Line and Stacked/Clustered Column  
  2. Drag first measure to **Column Values**  
  3. Drag second measure to **Line Values**  
  4. Set Axis (e.g., Product Category or Month)  
- **Example:** Revenue (column) vs Profit Margin (line) per Product Category

# ⭐ 3. Waterfall Chart
- **Purpose:** Show **cumulative effect** of sequential positive and negative values  
- **Steps to create:**  
  1. Insert → Waterfall Chart  
  2. Drag category (e.g., Product) to **Category**  
  3. Drag measure (e.g., Total Revenue) to **Y-axis**  
- **Example:** Revenue changes month by month

# ⭐ 4. Funnel Chart
- **Purpose:** Visualize **stages of a process** and drop-offs  
- **Steps to create:**  
  1. Insert → Funnel Chart  
  2. Drag Stage (e.g., Sales Stage) to **Group**  
  3. Drag measure (e.g., Leads or Revenue) to **Values**  
- **Example:** Leads → Opportunities → Closed Deals

# ⭐ 5. Gauge Chart
- **Purpose:** Show **progress toward a goal**  
- **Steps to create:**  
  1. Insert → Gauge  
  2. Drag measure (e.g., Total Revenue) to **Value**  
  3. Drag target measure (e.g., Target Revenue) to **Target value**  
- **Example:** Monthly revenue progress

# ⭐ 6. Map Visualizations
- **Map:** Visualize data by location  
  - Steps: Insert → Map → Location field → Value field  
  - Example: Sales by Country or City

- **Filled Map (Choropleth):** Color-coded regions  
  - Steps: Insert → Filled Map → Location → Value  
  - Example: Revenue per State/Region

- **ArcGIS Maps:** Advanced geographical analytics  
  - Requires ArcGIS subscription or Power BI Pro  
  - Adds layers, heatmaps, and spatial analysis

# ⭐ 7. Example Scenario
1. Combo chart: Revenue vs Profit Margin by Product Category  
2. Waterfall chart: Monthly revenue changes  
3. Funnel chart: Sales pipeline visualization  
4. Gauge chart: Monthly revenue progress toward target  
5. Map visual: Revenue by State  
6. Filled map: Color-coded revenue per region

# ⭐ 8. Key Takeaways
- Advanced visuals enhance **storytelling and complex analysis**  
- Combo charts compare multiple measures in one visual  
- Waterfall and funnel charts show **sequential changes or processes**  
- Gauge and KPI charts highlight **progress vs targets**  
- Map visuals provide **geospatial insights**

# ⭐ 9. Interview Questions

**Q1: What is a combo chart used for?**  
To show two different measures together, typically columns and a line.

**Q2: How does a waterfall chart work?**  
Displays sequential positive and negative changes to show cumulative effects.

**Q3: When should you use a funnel chart?**  
To visualize stages in a process and see drop-offs between stages.

**Q4: What is the purpose of a gauge chart?**  
To show progress toward a goal or target value.

**Q5: How do map visuals enhance reporting?**  
They provide geospatial analysis and location-based insights.

**Q6: What is the difference between a basic map and a filled map?**  
A filled map colors regions by values, while a basic map shows points for locations.

**Q7: Can you use advanced visuals for KPIs?**  
Yes, combo, waterfall, gauge, and funnel visuals are often used to highlight KPIs.

**Q8: What is ArcGIS map used for in Power BI?**  
For advanced spatial analysis, heatmaps, and layered geospatial insights.

**Q9: When would you use a waterfall chart over a line chart?**  
When you want to show step-by-step cumulative changes rather than trends.

**Q10: How do combo charts help in decision-making?**  
By comparing two metrics in a single visual, helping users identify relationships and patterns.
