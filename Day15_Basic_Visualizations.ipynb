# 📘 Day 15 — Basic Visualizations

# ⭐ 1. Overview
Day 15 focuses on **creating and understanding basic visuals in Power BI**, such as columns, bars, lines, pies, and tables.  
Visuals help represent data **clearly and interactively**.

# ⭐ 2. Common Visual Types

- **Clustered Column Chart:**  
  - Compare values across categories  
  - Example: Revenue per Product Category  
  - Steps: Insert → Column Chart → Axis: Product Category → Values: Total Revenue

- **Stacked Column Chart:**  
  - Compare parts of a whole across categories  
  - Example: Revenue by Product Category per Region  
  - Steps: Insert → Stacked Column Chart → Axis: Product Category → Legend: Region → Values: Total Revenue

- **Clustered Bar Chart:**  
  - Similar to column chart but horizontal  
  - Good for long category names

- **Line Chart:**  
  - Show trends over time  
  - Example: YTD Sales trend  
  - Steps: Insert → Line Chart → Axis: OrderDate → Values: Total Revenue

- **Area Chart:**  
  - Similar to line chart but filled area under the line  
  - Good for cumulative trends

- **Pie Chart:**  
  - Show proportion of a whole  
  - Example: Revenue share by Product Category  
  - Steps: Insert → Pie Chart → Legend: Product Category → Values: Total Revenue

- **Scatter Chart:**  
  - Compare two numeric values and see correlations  
  - Example: Quantity vs Total Price  
  - Steps: Insert → Scatter Chart → X Axis: Quantity → Y Axis: Total Price → Details: ProductID

- **Table Visual:**  
  - Display detailed data in tabular form  
  - Example: SalesID, Product, Quantity, Total Revenue  
  - Steps: Insert → Table → Add required columns

- **Matrix Visual:**  
  - Pivot table style; rows, columns, and values  
  - Example: Revenue by Product Category and Region  
  - Steps: Insert → Matrix → Rows: Product Category → Columns: Region → Values: Total Revenue

# ⭐ 3. Slicers & Filters

- **Slicer:**  
  - Interactive filter for visuals  
  - Example: Filter visuals by Year, Month, or Product  
  - Steps: Insert → Slicer → Field: Year/Month/Product → Apply to visuals

- **Filters Pane:**  
  - Filter data for report page or visual  
  - Steps: Select visual → Filters pane → Add field → Set filter conditions

# ⭐ 4. Example Scenario

1. Create **Column Chart** for Revenue by Product Category  
2. Create **Line Chart** for YTD Sales trend over months  
3. Create **Pie Chart** for Revenue distribution by Product Category  
4. Create **Table** for detailed Sales data (SalesID, Product, Quantity, Total Revenue)  
5. Add **Slicer** for Year and Product to filter all visuals  
6. Test interactions by selecting different slicer values

# ⭐ 5. Key Takeaways

- Choose visuals based on **data type and analysis goal**  
- Column and bar charts: comparisons  
- Line and area charts: trends  
- Pie charts: proportions  
- Table and matrix: detailed analysis  
- Slicers and filters: interactive dashboards

# ⭐ 6. Interview Questions

**Q1: What is a clustered column chart used for?**  
To compare values across different categories vertically.

**Q2: How is a stacked column chart different from a clustered column chart?**  
Stacked columns show parts of a whole for each category; clustered columns show side-by-side comparisons.

**Q3: When should you use a line chart?**  
To display trends over time.

**Q4: What is a matrix visual in Power BI?**  
A pivot-style table with rows, columns, and aggregated values.

**Q5: How do you create a slicer?**  
Insert → Slicer → Add field (e.g., Year, Product) → Apply to visuals.

**Q6: What is the difference between a table and a matrix visual?**  
Table displays data in rows and columns; matrix allows grouping and aggregation with rows and columns.

**Q7: When is a pie chart appropriate?**  
To show proportion of categories within a whole.

**Q8: What is a scatter chart used for?**  
To visualize correlation between two numeric variables.

**Q9: How do filters affect visuals in Power BI?**  
Filters restrict the data displayed in visuals, either at visual, page, or report level.

**Q10: Why are interactive slicers important in dashboards?**  
They allow users to dynamically filter and explore data in real time.
